```python
import pandas as pd
import numpy as np
```

```python
sales_data= {
    'Day' : ['Mon', 'Tues', 'Wed', 'Thurs', 'Fri', 'Sat', 'Sun'],
    'Sales' : [45,50,48,52,47,300,49]
}
df_sales=pd.DataFrame(sales_data)
print(df_sales)
```

```python
from numpy._core.defchararray import lower
Q1=df_sales['Sales'].quantile(0.25)
Q3=df_sales['Sales'].quantile(0.75)

IQR= Q3-Q1

lower_limit=Q1 -1.5 * IQR
upper_limit= Q3 + 1.5 * IQR

print('Lower Limit: ', lower_limit)
print('Upper Limit: ', upper_limit)

outliers = df_sales[(df_sales['Sales'] < lower_limit) | (df_sales['Sales'] > upper_limit)]
print("\nOutlier rows: ", outliers)
```

```python
#Fix oUTLIER BY CAPPING IT
df_sales['Sales_Capped'] =np.where(
    df_sales['Sales'] > upper_limit, upper_limit, df_sales['Sales']
)
print(df_sales)
```
