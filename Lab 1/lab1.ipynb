```python
import pandas as pd
import numpy as np

cafe_data ={
    'Cafe_Item' : ['Coffee', 'Tea', np.nan, 'Coffee', 'Cake'] ,
    'Daily_Sales' : [30,25,40, np.nan, 35] ,
    'Customer_Rating' : [4.5,4.0,4.8,4.2,np.nan]
}
dfcafe_data=pd.DataFrame(cafe_data)
#1. Which Cafe item will be used to replace the missing item?
#Common convention is mode so we will first use that

print("Filling missing values using mode")
Item_mode=dfcafe_data['Cafe_Item'].mode()[0]
dfcafe_data['Cafe_Item']=dfcafe_data['Cafe_Item'].fillna(Item_mode)
print("Missing Cafe Items: ",Item_mode)
```

```python
#2. Fill the missing Daily Sales using the Mean(Average) sales.
mean_sales=dfcafe_data['Daily_Sales'].mean()
dfcafe_data['Daily_Sales']=dfcafe_data['Daily_Sales'].fillna(mean_sales)
print("Average Daily Sales: ", mean_sales)
```

```python
mean_rating=dfcafe_data['Customer_Rating'].mean()
dfcafe_data['Customer_Rating']=dfcafe_data['Customer_Rating'].fillna(mean_rating)
print("Average Customer Rating: ", mean_rating)
```
