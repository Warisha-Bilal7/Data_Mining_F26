```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
```

```python
pharmacy_transactions = [
    ['Mask', 'Sanitizer'],
    ['Mask', 'Sanitizer', 'Vitamin'],
    ['Mask', 'Vitamin', 'Bandage'],
    ['Sanitizer', 'Bandage'],
    ['Mask', 'Sanitizer', 'Bandage'],
    ['Mask', 'Sanitizer', 'Vitamin', 'Bandage'],
    ['Sanitizer', 'Vitamin'],
    ['Mask', 'Bandage']
]
```

```python
te = TransactionEncoder()
te_ary = te.fit(pharmacy_transactions).transform(pharmacy_transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("--- Encoded DataFrame ---")
print(df)
print("\n" + "="*50 + "\n")
```

```python
# Task 02: Apriori Algorithm (min_support = 37.5% or 3/8 = 0.375)
min_support = 0.375
frequent_itemsets_apriori = apriori(df, min_support=min_support, use_colnames=True)

print("--- Frequent Itemsets (Apriori) ---")
print(frequent_itemsets_apriori)
print("\n" + "="*50 + "\n")
```

```python
# Task 03: FP-Growth Algorithm (min_support = 37.5%)
frequent_itemsets_fpgrowth = fpgrowth(df, min_support=min_support, use_colnames=True)

print("--- Frequent Itemsets (FP-Growth) ---")
print(frequent_itemsets_fpgrowth)
print("\n" + "="*50 + "\n")
```

```python
# Task 04: Generate Association Rules (min_confidence = 60% or 0.60)
rules = association_rules(frequent_itemsets_fpgrowth, metric="confidence", min_threshold=0.6)

# Selecting relevant columns for neat display
rules_display = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
print("--- Association Rules ---")
print(rules_display)
```

Q1: Which two items have the highest support, and what is that value?
Answer: Mask and Sanitizer both have the highest individual item support.Support Value: $\frac{6}{8} = 0.75$ (or 75%).Mask appears in 6 out of 8 transactions.Sanitizer appears in 6 out of 8 transactions.

Q2: Do Apriori and FP-Growth give the same frequent itemsets? Why do you think this is so?
Answer: Yes, both algorithms yield the exact same set of frequent itemsets.

Reason: Both algorithms implement the same mathematical definition of support threshold for itemsets. While Apriori uses a candidate-generation approach (generate-and-test) and FP-Growth uses a compressed FP-Tree structure to mine itemsets directly without candidate generation, the underlying problem statement and threshold filter are identical.

Q3: Which rule has the highest confidence, and what is its value?Answer: The rule {Vitamin} $\rightarrow$ {Mask} has the highest confidence.Confidence Value: $\frac{3}{4} = 0.75$ (or 75%).Calculation: Out of the 4 transactions containing Vitamin, 3 also contain Mask ($3/4 = 75\%$).

Q4: Which rules have a lift of exactly 1? What does this tell us about those items?Answer:{Mask} $\rightarrow$ {Sanitizer} (Confidence = 66.7%, Lift = 1.0){Sanitizer} $\rightarrow$ {Mask} (Confidence = 66.7%, Lift = 1.0)Explanation: A lift equal to 1 indicates that the antecedent and consequent are statistically independent. Buying a Mask provides no additional information about the likelihood of buying a Sanitizer; customers buy Sanitizer at its standard baseline frequency regardless of whether a Mask is in their basket.

Q5: The rule Mask $\rightarrow$ Sanitizer has 66.7% confidence. Is it a good rule? Explain using its lift.Answer: No, it is not a good rule.Explanation: Although 66.7% sounds moderately high for confidence, its Lift is 1.0. Since the baseline purchase rate of Sanitizer in the entire store is already 75% ($6/8$), observing a Mask actually corresponds to a lower proportion of Sanitizer purchases ($66.7\%$) compared to the general baseline. Because the lift is not greater than 1, the rule holds no predictive or actionable value for cross-selling.

Q6: Why is lift a better judge of a rule than confidence alone?Answer:Confidence only measures how often the consequent appears given the antecedent, ignoring how popular the consequent is overall. High confidence can be deceiving if the consequent is simply a very popular item bought by almost everyone regardless of the antecedent.Lift measures how much more likely the consequent is bought when the antecedent is present compared to its natural baseline popularity ($\text{Lift} = \frac{\text{Confidence}}{\text{Support(Consequent)}}$).$\text{Lift} > 1$: Strong positive co-occurrence (valuable rule).$\text{Lift} = 1$: Independent items (useless rule).$\text{Lift} < 1$: Negative co-occurrence / substitution effect.
